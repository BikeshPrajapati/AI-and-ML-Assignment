```python
import pandas as pd
import numpy as np
```

# Load Dataset

```python
df = pd.read_csv('LoanPrediction.csv')
```

```python
df.head()
```

```python
df.shape
```

```python
#Drop the Loan_ID column as it just an identifier
df.drop(columns=['Loan_ID'], inplace = True)
```

```python
df.shape
```

# Missing Values

```python
## Identify categorical columns in the dataframe
categorical_cols  = df.select_dtypes(include='object').columns
categorical_cols
```

```python
## Handle Categorical Missing Values

for col in categorical_cols:
    #print(col)
    #print(df[col].mode()
    mode = df[col].mode()[0] #most common/frequent value
    df[col] = df[col].fillna(mode) #Fill missing values with mode of that column
```

```python
## Handle Numerical Missing Values
numerical_cols  = df.select_dtypes(include='number').columns

for col in numerical_cols:
    median = df[col].median() #find median of that column
    df[col] = df[col].fillna(median) #Fill missing values with median of that column
```

```python
df.head()
```

# Bins 

```python
# Bin Applicant Income into 3 categories : Low, Medium, High

bins = [0, 2500, 6000, np.inf]
labels = ['Low', 'Medium', 'High']
#Create new column
df['Income_bin'] = pd.cut(df['ApplicantIncome'], bins = bins, labels = labels)
```

```python
#Check distribution of Income_bin
df['Income_bin'].value_counts()
```

```python
df.tail()
```

```python
#Print column data types
print(df.dtypes)
```

```python
#Convert the column from 'category' data type to 'object data type
df['Income_bin'] = df['Income_bin'].astype('object')
```

```python
#Print column data types
print(df.dtypes)
```

# Feature Creation

```python
# Create a new Feature : Debt to Income Ratio = LoanAmount / Applicant Income and Coapplicant Income

df['Debt_Income_Ratio'] =  df['LoanAmount']/(df['ApplicantIncome']+df['CoapplicantIncome']+1) # 1500/(0+0+1) 

#Total Income
df['Total_Income'] = df['ApplicantIncome']+df['CoapplicantIncome']
```

```python
df.head()
```

```python
#Print column data types
print(df.dtypes)
```

# Encoding Categorical Variables

```python
# One hot encoding of all categorical variables excluding target variable
target = 'Loan_Status'

#Identify categorical columns
categorical_cols  = df.select_dtypes(include='object').columns

#new_df = pd.get_dummies(df, columns=[col for col in categorical_cols if col != target])
df = pd.get_dummies(df, columns=[col for col in categorical_cols if col != target], drop_first=True) # multicollinearity 

# Encode target variable
# 'Y' -> 1, 'N' -> 0
df[target] = df[target].map({'Y': 1, 'N': 0})
```

```python
print(df.dtypes)
```

```python
# Only convert boolean columns to int
bool_cols = df.select_dtypes(include='bool').columns
df[bool_cols] = df[bool_cols].astype(int)
```

# Export Cleaned Dataset

```python
df.to_csv('Loan_Clean.csv', index=False)
```
