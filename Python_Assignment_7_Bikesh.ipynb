21. Create a class Bank Account with methonds:

```python
class BankAccount:
    balance = 0
    def __init__(self, balance= None):
        if balance is None:
            raise ValueError ("You must enter balance to open account")
        if not isinstance(balance, (int, float)):
            raise TypeError ("Amount must be in integer or float ")
        self.balance = balance
    def deposit(self, amount = None):
        if amount is None:
            return "Error: You must provide certain amount."
        if not isinstance(amount, (int, float)):
            return "Error : Amount must be in integer or float "
        if amount > 0:
            self.balance += amount
        return f"Your deposit amount  is : {amount} and your total balance is : {self.balance}"

    def withdraw(self, amount = None):
        if amount is None:
            return "Error: You must provide certain amount."
        if not isinstance(amount, (int, float)):
            return "Error : Amount must be in integer or float "
        if self.balance >= amount:
            self.balance -= amount
            return f"You withdraw this amount : {amount} and your remaining balance is : {self.balance}"
        else:
            return f"You Don't have enough money."

    def display_balance(self):
        return f"Your Total Balance is : {self.balance}"



BA1 = BankAccount(100)
BA1.deposit(100)
```

    - borrow()
    - return_book()
    - status()


```python
class LibraryBook:
    def __init__(self, *books):

        if not isinstance(books,tuple):
            raise TypeError("Please pass the titles of books")

        self.books = {title : True for title in books}

    def add_book(self, *titles):
        if not isinstance(titles, tuple):
            raise TypeError("Please pass the titles of books")
        for title in titles:
            self.books[title] = True

        # tracks what user has
    def borrow(self,current_book = None):

        if current_book is None:
            print("No book selected")
            return

        if current_book not in self.books:
            print("Book not available")
            return

        elif current_book in self.books:
            self.books[current_book] = False
        print(f"You borrowed '{current_book}'")

    def return_book(self, return_book = None):
        if return_book is None:
            print("No book selected")
            return

        if return_book not in self.books:
            print("You don't have this book")
            return
        elif return_book in self.books and self.books[return_book] is False:
            self.books[return_book] = True
            print(f"You returned '{return_book}'")

        else:
            print(f"This book is available to borrow: {return_book}")


    def status(self):
        print("\n--- STATUS ---")
        print("Available books:", [book for book,status in self.books.items() if status])
        print("Borrowed books:", [book for book,status in self.books.items() if not status])



```

25. Create a class Order with methonds:
    - add_items(price)
    - calculate_total()
    - apply_discount(percent)



```python
class Order:
    def __init__(self):
        self.items = {
            "Apple": {"qty": 10, "price": 1.50},
            "Orange": {"qty": 10, "price": 1.00},
            "Banana": {"qty": 10, "price": 0.50},
        }
        self.available_items = self.items.copy()
        self.order_items = {}
        self.total= 0

    def add_items(self, item=None, qty=None, price=None):
        if item is None or qty is None or price is None:
            return "you must provide items  and qty"
        if not isinstance(item, str):
            return "items must be a string"
        if not isinstance(qty, int):
            return "The Quantity must be integer"
        if not isinstance(price, float):
            return "The Price must be float"
        self.items[item] = {"qty": qty, "price": price}
        print(f"This {item} is added")
        return self.items

    def add_order(self, item=None, qty=None):
        if item is None or qty is None:
            return "you must provide items  and qty"
        if not isinstance(item, str):
            return "item must be a string"
        if not isinstance(qty, int):
            return "The Quantity must be integer"

        if item not in self.available_items.keys():
            print("Item not Found")
            return
        info = self.available_items[item]

        if info["qty"] < qty:
            print(" Not enough quantity available")
            print(f" {item}:{info["qty"]} is left")
        if info["qty"] > qty :
            info["qty"] = info["qty"] - qty
        if info["qty"] == qty :
            del self.available_items[item]

        print("--- Recently Added Order ---")
        print(f"{item}:{qty}")

        self.order_items[item] = {"qty": qty}

        return f"Available items are :{self.available_items}"

    def cal_total(self):

        for item, value in self.order_items.items():
            if item in self.available_items:
                unit_price = self.available_items[item]["price"]
                item_total = value["qty"] * unit_price
                self.total += item_total

                print(f"{item}: {value["qty"]} *{unit_price} = {item_total}")
            else:
                print(f"{item} is not available")

        print (f"Total order price : ${self.total}")
        return self.total

    def apply_discounts(self, percent = None):
        max_dis = 50
        if percent is None:
            return f"You dont have any discounts /n your total is {self.total}"
        if percent > max_dis or percent < 0:
            return "You have invalid discounts./n please provide the discount between 0 to 50"
        dis_amt =round(self.total *(percent/ 100),2)
        self.total -= round(dis_amt,2)

        print(f"your total discount is {dis_amt}")
        print(f"Your final total is {round(self.total,2)}")







order1 = Order()
```

Problem:
Design a class LRUCounter that:
- has a method hit(key)
- tracks how many times each key is accessed
- has a mehtond most_used() that returns the most acceessed key

- focus: state management inside class, methond logic, clean design

```python
class LRUCounter:
    def __init__ (self):
        self.counts = {}
        self.most_used_key = None
        self.most_used_count = 0


    def hit(self, key = None):
        if key is None:
            return "Please provide a key"
        if not isinstance(key, str):
            return "Please provide a string value"
        new_count = self.counts.get(key,0) + 1
        self.counts[key] = new_count

        if new_count > self.most_used_count:
            self.most_used_key = key
            self.most_used_count = new_count
        return self.counts

    def most_used(self):
        return self.most_used_key


LRU = LRUCounter()
LRU.hit("A")
LRU.hit("B")
LRU.hit("C")
LRU.hit("B")
LRU.hit("A")
LRU.hit("B")
LRU.hit("E")
```

#### MAANG Question 2 (Meta / Netflix - style)

Problem: Design a class Ratelimiter that:
- allows only N request per user
- has methond allow_request(used_id)
- returns True if allowed, False otherwise

- focus real-world object modeling  + methond responsibility


```python

class RateLimiter:
    def __init__(self, max_request):
        if max_request <= 0:
            raise ValueError("max_request must be positive")
        if not isinstance(max_request, int):
            raise ValueError("max_request must be an integer")
        self.user_request= {}
        self.max_request = max_request

    def allow_request(self, user_id= None):
        if user_id is None:
            return "Must provide user_id"
        if not isinstance(user_id, str):
            return  "Must provide a string value"

        count = self.user_request.get(user_id,0)

        if count >= self.max_request:
            return False

        self.user_request[user_id] = count+1
        return True



rate = RateLimiter(5)

rate.allow_request("tom")
```
