```python
import pandas as pd
from sklearn.preprocessing import MinMaxScaler, StandardScaler

# Sample data
df = pd.DataFrame({
    'Income': [20000, 40000, 60000, 100000],
    'LoanAmount': [100, 200, 300, 500]
})

# Initialize scalers
minmax = MinMaxScaler()
standard = StandardScaler()

# Apply scaling
df_minmax = pd.DataFrame(minmax.fit_transform(df), columns=df.columns)
df_standard = pd.DataFrame(standard.fit_transform(df), columns=df.columns)

print("Original Data:\n", df)
print("\nMinMax Scaled:\n", df_minmax)
print("\nStandard Scaled:\n", df_standard)
```
