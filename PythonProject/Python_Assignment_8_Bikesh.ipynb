11. Create a Calculator that handles invalid operator input.

```python
class Calculator:
    def __init__(self, *nums):
        try:
            self.validate_numbers(*nums)
            self.nums = nums
        except ValueError as ve:
            print (f"Input Error : {ve}")
            self.nums = ()

    @staticmethod
    def validate_numbers(*nums):
        if not all(isinstance(n,(int,float)) for n in nums):
            raise ValueError("All inputs must be int or float")

    def add(self):
        try:
            return sum(self.nums)
        except Exception as e:
            print(f"Addition Error: {e}")

    def subtract(self):
        try:
            result = self.nums[0]
            for n in self.nums[1:] :
                result -= n
            return result
        except Exception as e:
            print (f"Subtraction Error : {e}")

    def multiply(self):
        try:
            result = 1
            for n in self.nums:
                result *=n
            return result
        except Exception as e:
            print(f"Multipication Error : {e}")

    def divide(self):
        try:
            result = self.nums[0]
            for n in self.nums[1:]:
                if n == 0:
                    raise ZeroDivisionError ("Cannot divide by Zero")
                result /= n
            return result
        except ZeroDivisionError as ze:
            print(f"Division Error : {ze}")
        except Exception as e:
            print(f"Division Error : {e}")
```

```python
class LengthBelowEightError(Exception):
    pass
class EmptyStrError(Exception):
    pass
class SpaceError(Exception):
    pass

def create_password(password = None):
    try:
        if password is None:
            raise TypeError("Please provide a password")
        if not isinstance(password, str):
            raise ValueError("Password must be string")
        if password == "":
            raise EmptyStrError("Please provide a full string value")
        if len(password) < 8:
            raise LengthBelowEightError("Password must be at least 8 characters long")

        if " " in password:
            raise SpaceError("Please remove space from password")
        return password
    except LengthBelowEightError as e:
        print (f" Error : {e}")
    except ValueError :
        print ("Please provide a string")
    except TypeError :
        print("Please provide a password")
    except EmptyStrError as ese:
        print(f"Error: {ese}")
    except SpaceError as se:
        print(f"Error : {se}")

```

16. create a class bankAcount with private balance and methond deposit() and withdraw().

```python
class NegativeAmtError(Exception):
    pass

class BankAccount():

    def __init__(self):
        self.__balance = 0

    @staticmethod
    def validate_amt(amt):
        if amt is None:
            raise ValueError("Please provide amt to deposit")

        if not isinstance(amt, (int, float)):
            raise ValueError("Please provide int or float as amt")

        if amt <= 0 :
            raise NegativeAmtError("Amount must be positive and greater than zero")

        return amt


    def deposit (self, amt = None):
        try:
            amt = self.validate_amt(amt)
            self.__balance += amt
            return f"The {amt} is deposited into your account"
        except(ValueError, NegativeAmtError) as e:
            return f"Deposite Failed: {e}"
    def withdraw(self, amt=None):
        try:
            amt = self.validate_amt(amt)

            if amt > self.__balance:
                return "Withdrawal failed: Insufficient balance"

            self.__balance -= amt
            return f"{amt} has been withdrawn from your account"

        except (ValueError, NegativeAmtError) as e:
            return f"Withdrawal failed: {e}"

    def check_balance(self):
        return f"Your current balance is {self.__balance}"

```

17. Create a class Student with private marks and methods to update marks safely.

```python
class Student:
    def __init__(self, name):
        self.name = name
        self.__marks = {
            "English": 0,
            "Math" : 0,
            "Science" :0,
            "History" :0,
            "Social" :0
        }
        self.__student_info= {"Name": self.name,
                            "Subject": self.__marks}

    def get_marks(self, subject):
        # returns marks for a given subject safely with report
        if subject not in self.__student_info["Subject"]:
            raise ValueError(f"{subject} is not a valid subject")
        return self.student_info["Subject"][subject]

    def update_marks(self, subject, marks):
        if isinstance(subject,str):
            subject = (subject,)
        elif isinstance(subject,list):
            subject = tuple(subject)

        if isinstance(marks,(int,float)):
            marks = (marks,)
        elif isinstance(marks,list):
            marks = tuple(marks)
        if len(subject) != len(marks):
            raise ValueError ("Number of subject and marks must match")

        for sub, mark in zip(subject, marks):
            # validate subject
            if sub not in self.__marks:
                raise ValueError (f"{subject} is not valid subject")

            if not isinstance(mark, (int, float)):
                raise ValueError ("Marks must be number")
            if not 0 <= mark <= 100:
                raise ValueError ("Marks must be between 0 and 100")
            # update Safely
            self.__marks[sub] = mark
        return self.__marks

    def get_result(self):
        print (f"------ Result_______\n"
                f"Name: {self.__student_info['Name']} \n"
                f"Marks :{self.__student_info['Subject']}")
```

```python
class Employee:

    def __init__(self, name= None,jobTitle = None, department= None, salary = None ):
        if None in (name, jobTitle,department, salary):
            raise TypeError ("Name, jobtitle, Department must be provided")
        self.set_name(name)
        self.set_jobTitle(jobTitle)
        self.set_department(department)
        self.set_salary(salary)

        # Setter
    def set_name(self, name):
        if not isinstance(name, str) or not name.strip():
            raise ValueError ("Name must be full- string")
        self._name = name

    def set_jobTitle(self, jobTitle):
        if not isinstance(jobTitle, str) or not jobTitle.strip():
            raise ValueError ("Job Title must be full- string")
        self._jobTitle = jobTitle

    def set_department(self, department):
        if not isinstance(department, str) or not department.strip():
            raise ValueError ("Department must be full- string")
        self._department = department
    def set_salary(self, salary):
        if not isinstance(salary, (int, float)) or salary <= 0:
            raise ValueError("Salary must be a positive number")
        self._salary = salary


    # getter
    def get_name(self):
        return self._name
    def get_jobTitle(self):
        return self._jobTitle
    def get_department(self):
        return self._department
    def get_salary(self):
        return self._salary

    def employee_info(self):
        return f"{self._name},{self._jobTitle},{self._department},{self._salary}"


class Manager(Employee):

    def __init__(self, name = None,department = None, salary = None):

        super().__init__(
            name = name,
            jobTitle = "Manager",
            department= department,
            salary = salary)

    def update_employee(self,employee, name=None, jobTitle=None, department=None, salary = None):
        if not isinstance(employee, Employee):
            raise TypeError("You can only update Employee objects")
        if all(value is None for value in (name, jobTitle, department, salary)):
            raise ValueError ( "No updates were provided. Nothing was changed.")

        if name is not None:
            employee.set_name(name)
        if jobTitle is not None:
            employee.set_jobTitle(jobTitle)
        if department is not None:
            employee.set_department(department)

        return f"{employee.get_name()}'s information updated successfully."


```

```python
class Person:
    def __init__(self, name = None, address = None):
        if None in (name, address):
            raise TypeError("You must provide name and address")
        self.set_name(name)
        self.set_address(address)
    # setter
    def set_name(self,name):
        if not isinstance(name, str) or not name.strip():
            raise ValueError("Name must be string")
        self._name = name

    def set_address(self, address):
        if not isinstance(address,str) or not address.strip():
            raise ValueError ("Address must be string")
        self._address = address

    #getter
    def get_name(self):
        return self._name
    def get_address(self):
        return self._address
    def info(self):
        return (f"Name :{self._name}\n"
                f"Address :{self._address}")

class Teacher(Person):
    def __init__(self, name = None, address = None, title= None, salary = None):
        if None in (title,salary):
            raise TypeError("YOu must provide Name, title and salary")
        super().__init__(
            name,
            address
        )
        self.set_title(title)
        self.set_salary(salary)

    def set_title(self,title):
        if not isinstance(title, str) or not title.strip():
            raise ValueError("Title must be string")
        if title == "Principal":
            self._title = title
        else:
            self._title = f"{title}-Prof"

    def set_salary(self,salary):
        if not isinstance(salary, int) or not salary:
            raise ValueError("Salary must be integer")
        self._salary = salary

    def info(self):
        return (f"Name: {self._name}\n"
                f"Address: {self._address}\n"
                f"Title: {self._title}\n"
                f"Salary: {self._salary}")
    #setter


class Principal(Teacher):
    def __init__(self, name = None, address = None,  salary = None):
        super().__init__(
            name = name,
            address = address,
            title = "Principal",
            salary = salary
        )

    def update_teacher(self,teacher,name=None, address=None, title=None, salary=None):
        if not isinstance(teacher, Teacher):
            raise TypeError("You can only update teacher objects")
        if all(value is None for value in (name, address,title, salary)):
            raise ValueError ( "No updates were provided. Nothing was changed.")
        if name is not None:
            teacher.set_name(name)
        if address is not None:
            teacher.set_address(address)
        if title is not None:
            teacher.set_title(title)
        if salary is not None:
            teacher.set_salary(salary)

    def update_self(self, name=None, address=None, salary=None):
        if name is not None:
            self.set_name(name)
        if address is not None:
            self.set_address(address)
        if salary is not None:
            self.set_salary(salary)
    def info(self):
        return super().info()

```

```python
import csv

class CSVReader:

    def __init__(self, filePath = None,mode = None):
        if None in (filePath, mode):
            raise TypeError("You must provide filepath and mode to read files")
        self.set_filepath(filePath)
        self.set_mode (mode)

    #setter
    def set_filepath(self,filePath):
        if not isinstance(filePath, str):
            raise ValueError("filePath must be string")
        self._filePath = filePath

    def set_mode(self, mode):
        if not isinstance(mode,str):
            raise ValueError("Mode must be string")
        if mode not in ('r',):
            raise ValueError ("Invalid mode, Only 'r' is supported")
        self._mode = mode
    #getter
    def get_filepath(self):
        return self._filePath
    def get_mode(self):
        return self._mode

    def read_file(self):
        try:
            with open(self._filePath,mode = self._mode) as file:
                reader = csv.reader(file)
                self._data = list(reader)
        except FileNotFoundError as fe:
            return f"Err : {fe}"

        return self._data

    def get_grade(self):
        if not hasattr(self, "_data"):
            raise AttributeError("No data loaded. Call read_file() first.")

        return [row[2] for row in self._data if len(row) > 2]



```

29. Hangling missing Values in CSV using exception Hangling.

```python

class  CSVCleaner(CSVReader):
    def __init__(self, filepath = None, mode = "r"):
        super().__init__(
            filepath, mode
        )
        self._clean_data = []

    def clean_value(self,value, expected_type):
        # Replace  missing value using exception handling
        try:
            if expected_type == str:
                if value.strip() == "":
                    return "N/A"
                return value
            elif expected_type == int:
                return int(value)
            elif expected_type == float:
                return float(value)
        except (ValueError, TypeError) as e:
            # return default values when it fails
            if expected_type == str:
                return "N/A"
            elif expected_type == int:
                return 0
            elif expected_type == float:
                return 0.0

    def replace_missing_value(self, exp_type = None):
        raw_data = self.read_file()
        header = raw_data[0]
        num_cols = len(header)
        self._clean_data = [header]

        # if no types provided, default everything to string
        if exp_type is None:
            exp_type = [str] * num_cols

        for row in raw_data[1:]:
            clean_row = []
            for i in range (num_cols):
                value = row[i] if i < len(row) else ""
                clean_value = self.clean_value(value, exp_type[i])
                clean_row.append(clean_value)
            self._clean_data.append(clean_row)

        return self._clean_data

    def remove_rows(self):
        raw_data = self.read_file()
        header = raw_data[0]
        self._clean_data = [header]

        # this skips the rows that any values is missing
        for rows in raw_data[1:]:
            if all(cell.strip() for cell in rows):
                self._clean_data.append(rows)
        return self._clean_data
```

```python
import os

class TransactionProcessor(CSVFile):
    def __init__(self,filepath = None):
        super().__init__(filepath)

        # CASE 1: File exists and has content
        if os.path.exists(filepath):
            self.raw_data = self.read_file()

            self.raw_data = [row for row in self.raw_data if row]


            if len(self.raw_data) == 0:
                # File exists but empty → initialize structure
                self.header = ["ID", "Amount", "Type"]
                self.values = []
                self.write_file([self.header])
            else:
                # File has data
                self.header = self.raw_data[0]
                self.values = self.convert_rows(self.raw_data[1:])


        # CASE 2: File does NOT exist → create new file
        else:
            self.header = ["ID", "Amount", "Type"]
            self.values = []
            self.write_file([self.header])

        self.__balance = self.compute_balance()

    def convert_rows(self,rows):
        clean = []
        for row in rows:
            if len(row) <3:
                continue
            try:
                amt = float(row[1]) if "." in row[1] else int(row[1])
                clean.append([int(row[0]), amt, row[2]])
            except:
                continue
        return clean



    # validation for amount
    def amt_validation(self,amt):
       if amt is None:
            raise TypeError("Please provide amt to deposit")
       if not isinstance(amt, (int,float)):
            raise ValueError ("Amount must be int/ float" )
       if  amt <= 0 :
           raise ValueError ("Invalid Amount")
       return amt

    def deposit_amount(self, amt = None):
        amt = self.amt_validation(amt)
        new_id = len(self.values) +1
        rows = [new_id, amt, "Debit"]
        self.values.append(rows)
        self.__balance += amt
        self.append_file([rows])
        return "Deposit Successful!"


    def withdraw_amount(self, amt= None):
        amt = self.amt_validation(amt)
        if self.__balance == 0 or self.__balance < amt:
            return "Insufficient Balance!!!"
        new_id = len(self.values) + 1
        rows = [new_id, amt, "Credit"]
        self.values.append(rows)
        self.__balance -= amt

        self.append_file([rows])
        return "Withdraw Successful!"





    def compute_balance(self):
        debit = sum(x[1] for x in self.values if x[2] == "Debit")
        credit = sum(x[1] for x in self.values if x[2] == "Credit")
        return debit - credit

    def total_debit(self):
        total_d = sum(x[1] for x in self.values if x[2] == "Debit")
        return f"The total Debit : {total_d}"

    def total_credit(self):
        total_c = sum(x[1] for x in self.values if x[2] == "Credit")
        return f"The total Credit : {total_c}"

    def net_balance(self):
        return f"The net Balance is : {self.__balance}"


    def show_all_transactions(self):
        if len(self.values) == 0:
            return "No transactions found."

        print(f"{self.header[0]:<5} {self.header[1]:<10} {self.header[2]:<10}")
        print("-" * 30)

        for row in self.values:
            print(f"{row[0]:<5} {row[1]:<10} {row[2]:<10}")

        return "Displayed all transactions."


```

##### Polymorphic Notification System

Design a notificaiton system using polymorphism:

- base class : Notification
- Derived Classes:

        - Email Notification
        - SMSNotification
        - PushNotification

Each has method send(message).

write a fucntion  notify_all(list_of_notificatoin) that sends message to all.
Add exception handling if sending fails.
- This is real system design + OOP polymorphism



```python
class Notification:

    def send(self,message : str):
        raise NotImplementedError ("Subclass must implement send() method ")

class EmailNotification(Notification):
    def send(self,message :str):
        try:
            print (f"[EMAIL] {message}")

        except Exception as e:
            print(f"Email Noticiation error: {e}")

class SMSNotification(Notification):
    def send(self, message: str):
        try:
            print(f"[SMS] {message}")
        except Exception as e:
            print(f"SMSNotification error: {e}")

class PushNotification(Notification):
    def send(self, message: str):
        try:
            print(f"[PUSH] {message}")
        except Exception as e:
            print(f"PushNotification error: {e}")
```

```python
def notify_all(notifications, message: str):
    for notifier in notifications:
        notifier.send(message)

notifiers = [
    EmailNotification(),
    SMSNotification(),
    PushNotification()
]

notify_all(notifiers, "System maintenance at 11 PM.")
```
