28. Write a function that takes two numbers and returns their LCM.

```python
def get_lcm(a= None, b= None):
    if a is None or b is None:
        return "Error: You must provide numbers."
       # Type validation
    if not isinstance(a, int) or not isinstance(b, int):
        return "TypeError: Both inputs must be integers."

    # Zero edge case
    if a == 0 or b == 0:
        return f"The LCM of {a} and{b} is : 0"
        # By definition, LCM(0, b) = 0

    # Negative numbers allowed — LCM is always positive
    a, b = abs(a), abs(b)

    bigger = max(a, b)
    step = bigger

    limit = a * b  # LCM must be ≤ a*b

    while bigger <= limit:
        if bigger % a == 0 and bigger % b == 0:
            return f"The LCM of {a} and {b} is : {bigger}"
        bigger += step

    # If we somehow get here, something is wrong
    raise ValueError("LCM does not exist for the given inputs.")

num1 = int(input("Enter your first number"))
num2 = int(input("Enter your second number"))

get_lcm(num1, num2)


```

35. Write a  fucntion  that takes two numbers and returns their HCF using logic (not built- in).

```python
def get_hcf (a= None, b= None):
    if a is None or b is None:
        return "Error: You must provide two numbers."
    if not isinstance(a, (int, float)):
        return "Error: You must provide integers/floats."
    num1, num2 = a, b
    a,b = abs(a), abs(b)
    if a == 0 and b ==0 :
        return "Value Error: HCF id undefined for (0,0)."
    if a == 0 :
        return  f" The HCF of {a} and {b} is :{b}"
    if b == 0 :
        return f" The HCF of {a} and {b} is :{a}"
    while b != 0:
        a,b = b, a %b

    return f" The HCF of {num1} and {num2} is : {a}"
```
