```python
def process_data(operation, initial_value, *args, **kwargs):
    result = initial_value
    if operation == "addition":
        for items in args:
            result = result + items
    elif operation == "subtraction":
        for items in args:
            result = result - items
    elif operation == "multiplication":
        for items in args:
            result = result * items

    if "message" in kwargs:
        print(kwargs["message"], end="")

    return result


print(process_data("multiplication", 5, 6, 7, 8, 9, 10, message="Result :"))
```

```python
class Rectangle:
    def __init__(self, width = 5.0, height=10.0):
        self.width = width
        self.height = height

    def cal_perimeter(self):
        perimeter = 2 * (self.width + self.height)
        return perimeter

rec_one = Rectangle(7.5,12.0)

print(rec_one.cal_perimeter())

rec_two = Rectangle()
print(rec_two.cal_perimeter())
```

```python
class LRUCache:

    def __init__(self, capacity: int):
        if 1 <= capacity <= 3000:
            self.capacity = capacity
        else:
            raise ValueError ("Capacity must be positive")

        self.cache = {}
        self. recent_used_key = None


    def get(self, key: int):
        if key is None:
            return "Please provide a key"
        if not isinstance(key, int):
            return "Please provide a integer value"
        if key in self.cache.keys():
            return key
        else:
            return -1


    def put(self, key: int, value: int):
        if key is None:
            return "Please Provide  a key"

        if not isinstance(key, int):
            return "Please provide a integer value"

        if key in self.cache.keys():
            self.recent_used_key = key
        else:
            self.cache[key] = value
            self.recent_used_key = key
        if len(self.cache) > self.capacity:
            self.cache.pop(self.recent_used_key)
            self.cache[key] = value

        return None
```

```python
lRU = LRUCache(2)
```

```python
lRU.put(1,1)
```

```python
lRU.put(2,2)
```

```python
lRU.get(1)
```
