```python
def process_data(operation, initial_value, *args, **kwargs):
    result = initial_value
    if operation == "addition":
        for items in args:
            result = result + items
    elif operation == "subtraction":
        for items in args:
            result = result - items
    elif operation == "multiplication":
        for items in args:
            result = result * items

    if "message" in kwargs:
        print(kwargs["message"], end="")

    return result


print(process_data("multiplication", 5, 6, 7, 8, 9, 10, message="Result :"))
```

```python
class LRUCache:

    def __init__(self, capacity: int):
        if 1 <= capacity <= 3000:
            self.capacity = capacity
        else:
            raise ValueError ("Capacity must be positive")

        self.cache = {}
        self. recent_used_key = None


    def get(self, key: int):
        if key is None:
            return "Please provide a key"
        if not isinstance(key, int):
            return "Please provide a integer value"
        if key in self.cache.keys():
            return key
        else:
            return -1


    def put(self, key: int, value: int):
        if key is None:
            return "Please Provide  a key"

        if not isinstance(key, int):
            return "Please provide a integer value"

        if key in self.cache.keys():
            self.recent_used_key = key
        else:
            self.cache[key] = value
            self.recent_used_key = key
        if len(self.cache) > self.capacity:
            self.cache.pop(self.recent_used_key)
            self.cache[key] = value

        return None
```

```python
lRU = LRUCache(2)
```

```python
import math
class Rectangle:
    def __init__(self, length, width):
        self.length = length
        self.width = width

    def calculate_area(self):
        area = self.length* self.width
        return area

class Circle:
    def __init__(self, radius):
        self.radius = radius

    def calculate_area(self):
        area = math.pi * (self.radius ** 2)
        return area

class Triangle:
    def __init__(self, base, height):
        self.base = base
        self.height = height

    def calculate_area(self):
        area= 0.5 * (self.base * self.height)
        return area



def print_area(shape):
    return shape.calculate_area()

print_area(shape = Rectangle(5,5))
print_area(shape = Triangle(5,5))
print_area(shape = Circle(5))


```

```python
class LengthBelowSixError(Exception):
    pass
class EmptyStrError(Exception):
    pass
class SpaceError(Exception):
    pass

def create_password(password = None):
    try:
        if len(password) < 6:
            raise LengthBelowSixError("Password must be at least 6 characters long")
        if not isinstance(password, str):
            raise ValueError("Password must be string")
        if password is None:
            raise TypeError("Please provide a password")
        if password == "":
            raise EmptyStrError("Please provide a full string value")
        if " " in password:
            raise SpaceError("Please remove space from password")
        return password
    except LengthBelowSixError:
        print ("Please enter a valid password")
    except ValueError:
        print ("Please provide all string values")
    except TypeError:
        print("Function does not take empty value")
    except EmptyStrError:
        print("Function does not take empty string")
    except SpaceError:
        print("Password cannot contain spaces")

```

```python
def find_hcf(a, b):
    while b != 0:
        a, b = b, a % b
    return a

def find_lcm(a, b):
    hcf = find_hcf(a, b)
    return (a * b) // hcf

# Example
a = 48
b = 18

print("HCF:", find_hcf(a, b))
print("LCM:", find_lcm(a, b))
```
