```python
# covert scores to grades
def covert_to_grade(scores):
    if scores >= 90 :
        return 'A'
    elif 80 <= scores <= 89:
        return 'B'
    elif 70 <= scores <= 79:
        return 'C'
    elif 50 <= scores <= 69:
        return 'D'
    return 'F'

scores.apply(covert_to_grade)





```

```python
import pandas as pd
data = {
    "Region" : ["North", "South","North","South","East"],
    "Category" : ["A","B","A","B","A"],
    "Sales": [102,145,123,145,150],
    "Profit" :[50,55,50,70,85]

}

df = pd.DataFrame(data)
```

```python
df
```

```python
agg_dict = {
    "Sales" : ["sum"],
    "Profit" :["mean"]}
df_sales_by_region = df.groupby(["Region"]).agg(agg_dict)
```

```python
df_sales_by_region
```
