```python
#Question 2

scores = pd.Series([75,82,91,68,87,90], index =['Alice','Bob','Charlie','David','Eve','Frank'])

print(scores[scores > 80])
print(scores[scores>=70].nunique())
```

```python
# covert scores to grades
def covert_to_grade(scores):
    if scores >= 90 :
        return 'A'
    elif 80 <= scores <= 89:
        return 'B'
    elif 70 <= scores <= 79:
        return 'C'
    elif 50 <= scores <= 69:
        return 'D'
    return 'F'

scores.apply(covert_to_grade)





```
