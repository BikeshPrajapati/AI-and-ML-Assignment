```python
# Seaborn Histogram

# distribution of conitniios variable based on bins - ranges
# detect belll curve -> Normal distribution
# Skeweness (left_skewed or right_skewed)
# outliter or ususal gaps
#helpful for EDA
iris = pd.read_csv('iris.csv')
sns.set_style("whitegrid")
sns.histplot(data = iris, x ='petal_length' )
```

```python
iris.columns
```

```python
import matplotlib.pyplot as plt

target ='species'
feature_cols = [col for col in iris.columns if col != target]

for col in feature_cols:
    plt.figure(figsize=(6,4))
    iris[col].hist(bins=20)
    plt.title(f"Histogram of {col}")
    plt.xlabel(col)
    plt.ylabel("Frequency")
    plt.show()
```

```python
for col in feature_cols:
    plt.figure(figsize=(6,4))
    sns.histplot(iris[col], kde=True, bins=30, color='blue')
    plt.title(f"Histogram of {col}")
    plt.xlabel(col)
    plt.ylabel("Count")
    plt.show()
```
