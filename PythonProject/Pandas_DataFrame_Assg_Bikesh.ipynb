##### Basic /filtering and new columns

```python
# 6 filter all customers whos premium_amount > 700
df.loc[df["premium_amount"] >700].head(10)
```

```python
# 7 Find Customer who have made at least 1 claim.
df.loc[df["claim_amount"]>0, "name"]
```

```python
#8. Add new col premium per year

df["premium_per_year"] = (df["premium_amount"] / df["tenure_years"]).round(2)
```

```python
df.head()
```

```python
# 9 create a new colum hascalim which shows Yes if calaim anout >0 else no
df["has_claim"] = (df["claim_amount"] > 0).map({True: "Yes", False: "No"})
```

```python
df.head()
```

```python
# 10. Filter customers who have health insurance nad live in albterta.

filters_rows = (df['policy_type']== 'health') &(df["region"]== "Alberta")
df.loc[filters_rows,["name","region","policy_type"]].head()
```

#### Medium(Agg, grouping, Renaming)

```python
#11 Find the Avg premium amount by region.

df.groupby('region')['premium_amount'].agg(["mean"])
```

```python
#12 find the total claim amount for each policytype.

df.groupby('policy_type')['claim_amount'].agg(["sum"])
```

```python
#13 calculate the maximum tenure yeras for each gender.

df.groupby('gender')['tenure_years'].agg(["max"])
```

```python
# 14 rename the column Premuim anount to premium-cad

df.rename(columns= {
    "premium_amount" :"premium_CAD"
}, inplace = True
)
```

```python
df.head()
```

```python
#15 Get the sum of premium amout and calim amt grouped by policty #type and region.
group_cols = ['policy_type','region']
cal_cols =['premium_CAD','claim_amount']
df.groupby(group_cols)[cal_cols].agg(["sum"])
```

#### Advanced/ Multi -condition amd computation

```python
#16 find customers whos premium amoutn is above the age for their policty type

df.loc[(df['premium_CAD'] > df.groupby('policy_type')['premium_CAD'].transform("mean")) ,"name"]
```

```python
# 17 Find the top 3 customers with the highest Claim AMount
df_sorted = df.sort_values(by='claim_amount', ascending=False)
```

```python
top3 = df_sorted["name"].head(3)
top3
```

```python
#18 Create a new column risk level
# "high " if calim amount >2000
#"Meidium" if between 500 -2000
# "Low" otherwise
df["risk_level"] = df["claim_amount"].apply(
    lambda x: "High" if x > 2000
    else "Medium" if 500 <= x <= 2000
    else "Low"
)
```

```python
df.head()
```

```python
#19 Filter cusotmer where
# tenure yeras >5 and
# no of calims > = 1 and
# policey type   ='life'


filter_conditons = (df['tenure_years'] >5) & (df['no_of_claims'] >= 1) & (df['policy_type'] == 'life')

df.loc[filter_conditons,["name","tenure_years","no_of_claims","policy_type"]]
```

```python
#20 find whih region has the highest total premium amount overall.

region_totals = df.groupby("region")["premium_CAD"].sum().sort_values(ascending=False)

region_totals.head(1)
```
