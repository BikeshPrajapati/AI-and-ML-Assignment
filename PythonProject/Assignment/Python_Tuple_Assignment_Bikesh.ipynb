10.Sorting Tuples Since tuples are immutable, they don't have a .sort() method. How would you create a sorted version of unsorted_tuple = (5, 1, 8, 2)? What is the data type of the result?

```python
unsorted_tuple = (5, 1, 8, 2)

sorted_tuple = sorted(unsorted_tuple)
```

```python

print(type(sorted_tuple))
```

```python
# to change back to tuple we need to wrap it:
sorted_tuple = tuple(sorted(unsorted_tuple))
```

```python
print(type(sorted_tuple))
```
