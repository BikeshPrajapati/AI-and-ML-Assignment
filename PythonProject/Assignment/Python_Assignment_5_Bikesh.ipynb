9. Write a program to print all the elements of a list a value 0 is found , using break.

```python
elements = [1,2,3,6,4,0,6,4]
i = 0
if not elements:
    print("Empty List")
else:
    while i < len(elements):
         if elements[i] == 0 :
            break
         else:
            print(elements[i])
            i += 1
```

```python
numbers = [3,6,2,7,3,7,4,9,67,34,13]
i = 0
odd_number = []
if not numbers:
    print("Empty List")
else:
    while i < len(numbers):
        if numbers[i] % 2 != 0 :
            odd_number.append(numbers[i])
        i += 1
print(f"The odd number is : {odd_number}")


```

13. Write a program to sum all elements of list using a whle loop

```python
numbers = [3,6,1,34,56,23,54]
i = 0
if not numbers:
    print("Empty List")
else:
    while i < len(numbers):
        sum = numbers[i] + numbers[i-1]
        i += 1

print(f"The sum is : {sum}")
```

18. Write a program to count how many even and odd numbers are present in list.

```python
numbers = [1,2,10,11,12,13,43,56,23,76,98,23,56]

if not numbers:
    print("Empty list")
else:
    even_counts = len([i for i in numbers if i%2 == 0])
    print(f"The count of even numbers is : {even_counts}")
    odd_counts = len([i for i in numbers if i%2 == 1])
    print(f"The count of odd numbers is : {odd_counts}")
```

23. Write a program to print duplicate elements form a list:

```python
from collections import Counter
numbers = [1,2,3,4,4,5,6,7,7,8,9,9]
if not numbers:
    print("Empty list")
else:
    counts= Counter(numbers)
    duplicate_list =[item for item, value in counts.items() if value >1]
    print(duplicate_list)
```
